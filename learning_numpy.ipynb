 1D array
 

```python
import numpy as np
arr=np.array([1,2,3,4])
print(arr)
print(type(arr))


```

```python
#reshaping one dimensional array
arr2=np.array([1,2,3,4,5])
arr2.reshape(1,5)
```

2D array

```python
arr3=np.array([[1,2,3,4],[5,6,8,9]])
print(arr3)
print(arr3.shape)
np.arange(0,10)#means elements in between 0 to 10
np.arange(0,8,3)
np.reshape(5,1)#5 row 1 column
```

```python
arr4=np.array([[1,2,3,4],[5,6,8,9]])
new_arr=np.reshape(arr4,(4,2))
print(new_arr)
```

Inbuilds function in numpy array


```python
np.ones((3,5))#used to create two dimension array of 3 rows and 5 columns
```

```python
np.eye(4)#identity matrix
```

```python
##Attributes of numpy array
arr=np.array([[1,2,3,4,5],[2,3,4,5,5]])
print("number of dimension",arr.ndim)
print("size of elements",arr.size)
print("data types",arr.dtype)
print("item size in bytes",arr.itemsize)
```

Numpy vectorized operation


```python
arr1=np.array([1,2,3,4,5])
arr2=np.array([10,20,30,40,50])
print("addition",arr1 + arr2)
print("subtractin",arr1 - arr2)
print("multiplication",arr1 * arr2)
print("division",arr1 / arr2)
```

```python
####universal array
arr=np.array([10,20,30])
print(np.sin(arr))
print(np.cos(arr))
print(np.sqrt(arr))
print(np.exp(arr))
print(np.log(arr))
```

```python
###array slicing
arr=np.array([[1,2,3,4,5],[5,6,7,8,9]])
print(arr)
print(arr[1:,3:])#row,column
print(arr[0][1])
print(arr[0:2,2:])#start at row 0 and end at row 2,column
```

```python
##Modify array elements
arr[0,1]=56
print(arr)
```

```python
arr[1:]=20
print(arr)
```

```python
###Normalization 
##to have a mean of 0 and standard deviation of 1
data=np.array([10,22,33,44])
mean=np.mean(data)
standard_dev=np.std(data)
normalized_data=(data-mean)/standard_dev
print(normalized_data)
```

```python
###statistical concept in numpy
data=([1,2,3,4,6,])
mean=print(np.mean(data))
standard_dev=print(np.std(data))
median=print(np.median(data))




```

```python
##logical operations
data=np.array([1,2,3,4,5,55,6])
print(data>5)
print(data[(data > 5) & (data < 8)])#and doesn't work
```
