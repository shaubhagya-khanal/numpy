```python
import pandas as pd
data = [1, 2, 3, 4, 5]
data = pd.Series(data)
print(data)
print(type(data))
```

```python
###creating a series from dictionary
##series is a one dimensional array like object that can hold any data types.It is similar to a column in a table
data={'a':1,'b':2,'c':3}
series_dict=pd.Series(data)
print(series_dict)
```

```python
#can be written this way too
data=[10,20,30]
index=['a','b','c']
pd.Series(data,index=index)
```

```python
##Dataframe
data={
    'Name':['krish','john','jack'],
    'Age':[25,34,55],
    'city':['bangalore','nweyork','florida']
}
print(data)
print(pd.DataFrame(data))
```

```python
##creating a data frame from a list of dictionaries
data=[
    {'Name':'krishna','Age':32,'city':'damauli'},
    {'Name':'adhyaya','Age':36,'city':'rampur'},
    {'Name':'akrit','Age':45,'city':'dolakha'}
     
]
print(pd.DataFrame(data))
```
