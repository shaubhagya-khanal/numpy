```python
import matplotlib.pyplot as plt
x=[1,2,3,4,5]
y=[3,4,5,7,22]
##creating a line plot
plt.plot(x,y)
plt.xlabel('x axis')
plt.ylabel('y axis')
plt.title('basic line plot')
plt.show()
 
```

```python
x=[1,2,3,4,5]
y=[2,3,4,5,6]
###creating a customized line plot
plt.plot(x,y,color='red',linestyle='--',marker='o',linewidth=3,markersize=22)
plt
plt
```

```python
import matplotlib.pyplot as plt

x = [1,2,3,4]
y1 = [2,4,6,8]
y2 = [1,3,5,7]

plt.figure(figsize=(11,6))

plt.subplot(2,2,1)
plt.plot(x, y1,color='blue')
plt.title("First")

plt.subplot(2,2,4)
plt.plot(x, y2)
plt.title("Second")

plt.subplot(2,2,3)
plt.plot(y1, y2)
plt.title("third")

plt.show()
```

```python
#bar plot
categories=['A','B','C','D']
values=[5,6,7,8]
plt.bar(categories,values,color='red')
```

```python
data=[1,5,3,4,5,8,6,7,8,4,3,6,]
plt.hist(data,bins=5,color='pink',edgecolor='black')#means it takes the data and divides into 5 bins
```

```python
###scatter plot
x=[1,2,3,4,4]
y=[2,5,4,5,7]
plt.scatter(x,y,marker="x")
```

```python
###piechart
labels=['A','B','C','D']
size=[20,30,40,45]
explode=(0.1,0,0,0)
colors=('red','blue','pink','grey')
plt.pie(size,labels=labels,explode=explode,colors=colors,autopct=('%1.1f%%'),shadow='True')
```

```python
import pandas as pd
df=pd.read_csv('sales_data.csv')
(df.head(5))
```

```python
##plot total sales bby product
total_sales_by_product=df.groupby('Product Category')['Total Revenue'].sum()
print(total_sales_by_product)
```

```python
plt.bar(total_sales_by_product.index,
        total_sales_by_product.values)
```

```python
#or
total_sales_by_product.plot(kind='bar',color='red')
```

```python
#or
total_sales_by_product = (
    df.groupby('Product Category')['Total Revenue']
      .sum()
      .reset_index()#now this becomes data frame by using reset index
)
plt.plot(
    total_sales_by_product['Product Category'],
    total_sales_by_product['Total Revenue']
)
```

in series we use index and value but in data frame we use column name(after reset_index)
